# Python implementations of the pseudocode in Algorithms Illuminated by Tim Roughgarden (Asymptotic Notation)

## Searching One Array - $O(n)$

```python
def search_element_arr(x, arr):
    return [True if x == element else False for element in arr]
```

```python
import big_o

gen = lambda n: (n - 1, list(range(n)))         

best, others = big_o.big_o(
    lambda args: search_element_arr(*args), gen,
    n_repeats=10, max_n=50_000,
)
print(big_o.reports.big_o_report(best, others))
```

## Searching Two Arrays - $O(n)$

```python
def search_two_arr(t, arr1, arr2):
    for x in arr1:
        if x == t:
            return True
    for y in arr2:
        if y == t:
            return True
    else:
        return False
```

```python
gen = lambda n: (n-1, list(range(n)), list(range(n)))

best, others = big_o.big_o(lambda args: search_two_arr(*args), gen, n_repeats=10, max_n=50_000)

print(big_o.reports.big_o_report(best, others))
```

## Checking for a Common Element - $O(n^2)$

```python
def check_comm_elem(arr1, arr2):
    for elem1 in arr1:
        for elem2 in arr2:
            if elem1 == elem2:
                return True
    return False
```

```python
gen = lambda n: (list(range(n)), list(range(n, 2 * n)))
best, others = big_o.big_o(lambda args: check_comm_elem(*args), gen, n_repeats=20, max_n=800)
print(big_o.reports.big_o_report(best, others))
```

## Checking for Duplicates - $O(n^2)$

```python
def duplicate_check(arr):
    i = 0
    while i < len(arr)-1:
        j = i + 1
        while j < len(arr):
            if arr[i] == arr[j]:
                return True
            j+=1
        i+=1
    return False
```

```python
gen = lambda n: list(range(n))
best, others = big_o.big_o(duplicate_check, gen, n_repeats = 20, max_n =800)
print(big_o.reports.big_o_report(best,others))
```
