# Python implementations of the pseudocode in Algorithms Illuminated by Tim Roughgarden (QuickSort)

## QuickSort - Naive first element pivot solution ($O(n^2)$)

```python
def quicksort(arr, l=0, r=None):

    if r is None: r = len(arr) - 1

    if l>=r: return arr 
    i = choose_pivot(arr, l, r)
    arr[l],arr[i]=arr[i],arr[l]
    j = partition(arr,l,r)
    quicksort(arr,l,j-1)
    quicksort(arr,j+1,r)
    return arr

def partition(arr,l,r):
    piv = arr[l]
    i, j = l+1, l+1
    for j in range(l+1, r+1):
        if arr[j] < piv:
            arr[i], arr[j] = arr[j], arr[i]
            i+=1
    arr[l],arr[i-1]=arr[i-1],arr[l]
    return i-1

def choose_pivot(arr,l,r):
    return l
```

```python
import numpy as np
import big_o

def rand_arr(n):
    return np.random.default_rng(42).choice(100000, size=n, replace=False).tolist()

best, others = big_o.big_o(lambda a: quicksort(a.copy()), rand_arr, min_n=100, max_n=100000, n_repeats=20)

print(big_o.reports.big_o_report(best, others))
```

## Randomised QuickSort - average case $O(n\log n)$

```python
import random

# quicksort and partition remain the same

def choose_pivot(arr, l, r):
    return random.randint(l,r)
```

# Non-comparison-based algorithms

## BucketSort - $O(n)$

```python
# Assuming the float data is in the range [0, 1]

def bucket_sort(arr, num_buckets=10):
    if len(arr) == 0:
        return []
    
    buckets = [[] for i in range(num_buckets)]
    for num in arr:
        if num < 0 or num > 1:
            raise ValueError("All numbers must be in the range [0, 1]")
        idx = min(num_buckets - 1, int(num * num_buckets))
        buckets[idx].append(num)
    out = []

    for bucket in buckets:
        for num in bucket:
            out.extend(sorted(num))
    return out
```

## CountingSort - $O(n)$

```python
def counting_sort(arr):
    if not arr:
        return []
    lo, hi = min(arr), max(arr)
    buckets = [[] for i in range(lo, hi+1)]
    for num in arr:
        buckets[num - lo].append(num)
    out = []
    
    for bucket in buckets:
        for num in bucket:
            out.append(num)
    return out
```
