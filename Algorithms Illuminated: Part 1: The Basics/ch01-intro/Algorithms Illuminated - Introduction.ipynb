## Standard School Multiplication 

```python
def standard_multiplication(x,y):

    result = 0
    
    for i in range(len(str(y))):
        current_y_digit = y//10**i%10
        addition_row = []
        for j in range(len(str(x))):
            current_x_digit = x//10**j%10
            addition_row_counter = -j-1

            if len(addition_row) < -addition_row_counter:

                if(current_x_digit*current_y_digit)<10:
                    addition_row.insert(addition_row_counter, current_x_digit*current_y_digit)
                elif(current_x_digit*current_y_digit)>=10:
                    addition_row.insert(addition_row_counter, current_x_digit*current_y_digit%10)
                    addition_row.insert(addition_row_counter-1,current_x_digit*current_y_digit//10)

            else:
                if current_x_digit*current_y_digit >= 10:
                    upd_value_cur_counter = current_x_digit*current_y_digit%10+addition_row[addition_row_counter]
                    if upd_value_cur_counter >= 10:
                        addition_row[addition_row_counter] = upd_value_cur_counter%10
                        addition_row.insert(addition_row_counter,upd_value_cur_counter//10+current_x_digit*current_y_digit//10)
                    else:
                        addition_row[addition_row_counter] = upd_value_cur_counter
                        addition_row.insert(addition_row_counter,current_x_digit*current_y_digit//10)
                else:
                    upd_value_cur_counter = current_x_digit*current_y_digit+addition_row[addition_row_counter]
                    if upd_value_cur_counter >= 10:
                        addition_row[addition_row_counter] = upd_value_cur_counter%10
                        addition_row.insert(addition_row_counter,upd_value_cur_counter//10)
                    else:
                        addition_row[addition_row_counter] = upd_value_cur_counter


        result += int("".join(map(str, addition_row)))*10**i

    return result


```

## Recursive Multiplication

```python
def recursive_multiplication(x,y):

    n = len(str(x))
    
    if n == 1:
        return x*y
    else:
        divider_x = 10**(n//2)
        divider_y = 10**(n//2)
        a = x // divider_x
        b = x % divider_x
        c = y // divider_y
        d = y % divider_y

        ac=recursive_multiplication(a,c)
        ad=recursive_multiplication(a,d)
        bc=recursive_multiplication(b,c)
        bd=recursive_multiplication(b,d)        

        return 10**(2*(n//2))*ac+10**(n//2)*(ad+bc)+bd

```

## Karatsuba Multiplication

```python
def karatsuba_multiplication(x,y):

    n = len(str(x))

    if n == 1:
        return x*y
    else:
        divider_x = 10**(n//2)
        divider_y = 10**(n//2)

        a = x // divider_x
        b = x % divider_x
        c = y // divider_y
        d = y % divider_y
        p = a + b
        q = c + d

        ac=recursive_multiplication(a,c)
        bd=recursive_multiplication(b,d)
        pq = recursive_multiplication(p,q)

        sum_ad_bc = pq-ac-bd
        return 10**(2*(n//2))*ac+10**(n/2)*sum_ad_bc+bd

        
```

### Validity tests

```python
import random 

for i in range(1000):
    x,y = random.randint(0, 999999), random.randint(0, 999999)
    assert standard_multiplication(x,y)==x*y, (x,y)

    n = random.choice([2,4,6])
    x,y = random.randint(10**(n-1), 10**n-1), random.randint(10**(n-1), 10**n -1)
    assert recursive_multiplication(x,y)==x*y, (x,y)
    assert karatsuba_multiplication(x,y)==x*y, (x,y)
print("Test complete")
```

### Time tests

```python
%timeit standard_multiplication(5678,1234)
%timeit recursive_multiplication(5678,1234)
%timeit karatsuba_multiplication(5678,1234) 
''' Having one less recursion step, Karatsuba trades one multiplication for several additions and substrations.
    This is faster for big numbers and slower for smaller ones.                                                '''
%timeit 5678*1234 # python uses native cpu multiplication and karatsuba multiplication method written on C for big integers
```

## MergeSort

```python
def mergesort(arr): 
    if len(arr) <= 1:
        return arr
    else:
        first_arr = mergesort(arr[0:len(arr)//2])
        second_arr = mergesort(arr[len(arr)//2:len(arr)])
        arr = merge(first_arr, second_arr)
    return arr

def merge(first_arr, second_arr):
    i,j=0,0
    output_arr=[] 
    while i < len(first_arr) and j < len(second_arr):
        if first_arr[i] < second_arr[j]:
            output_arr.append(first_arr[i])
            i+=1
        else:
            output_arr.append(second_arr[j])
            j+=1
    output_arr.extend(first_arr[i:])
    output_arr.extend(second_arr[j:])
    return output_arr
```

### Validity test

```python
import random 

check_index = 0
for i in range(0,9999):
    n = random.randint(0,999)
    array = random.sample(range(0, 9999), n)
    if check_index < 3:
        print(f"Array size: {n}. Array: {array}")
        check_index += 1
    assert mergesort(array) == sorted(array), array
```
