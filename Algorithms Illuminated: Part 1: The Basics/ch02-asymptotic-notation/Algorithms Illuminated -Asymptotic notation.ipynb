# # Python implementations of the pseudocode in Algorithms Illuminated by Tim Roughgarden (Asymptotic Notation)

## Searching One Array

```python
def search_element_arr(x, arr):
    return [x if x == element else 0 for element in arr]
```

```python
import big_o

gen = lambda n: (n - 1, list(range(n)))         

best, others = big_o.big_o(
    lambda args: search_element_arr(*args), gen,
    n_repeats=10, max_n=50_000,
)
print(big_o.reports.big_o_report(best, others))
```
