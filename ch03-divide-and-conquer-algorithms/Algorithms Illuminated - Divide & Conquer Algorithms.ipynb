## Brute Force Search for Counting Inversions - $O(n^2)$

```python
def count_invers_brute_force(arr):
    invers_num = 0
    for i in range(len(arr)-1):
        for j in range(i+1, len(arr)):
            if arr[i] > arr[j]:
                invers_num += 1
    return invers_num
```

```python
import big_o

gen = lambda n: (list(range(n, 0, -1)))

best, others = big_o.big_o(count_invers_brute_force, gen, n_repeats = 20, max_n = 800)

print(big_o.reports.big_o_report(best, others))
```

## Counting Inversions in $O(n\log n)$ Time

```python
def count_inversions(arr):
    if len(arr) <= 1:
        return (arr, 0)
    else:
        mid = len(arr)//2
        left_val, left_count = count_inversions(arr[:mid])
        right_val, right_count = count_inversions(arr[mid:])

        invers_num = left_count + right_count 

        merged, i, j = [], 0, 0
        while i < len(left_val) and j < len(right_val):
            if left_val[i] <= right_val[j]:
                merged.append(left_val[i])
                i+=1
            else:
                merged.append(right_val[i])
                j+=1
                invers_num += len(left_val) - i
```
